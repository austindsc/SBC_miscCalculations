```python
def fermiArg(K):
    
    gam = gamma(m_e, K)
    beta = get_beta(m_e, K)
    eta = gam*beta
    
    return np.sqrt(1 + eta**2)/eta

def paperArg(K):
    W = K/m_e + 1
    gam = gamma(m_e, K)
    beta = get_beta(m_e, K)
    P = gam*beta
    
    return W/P

getP(m_e, 1160*keV)/m_e
```
