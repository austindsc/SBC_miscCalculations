```python
import sys
import numpy as np
import matplotlib.pyplot as plt

pi = np.pi
```

```python
def thomsonCS(theta):
    return (1+np.cos(theta*pi/180)**2)/2

def thomsonE(theta, E, M):
    p = E
    return 2*(p*np.sin(theta * pi/180 /2))**2/M
               
def maxEr(E, M):
    return thomsonE(theta = 180, E = E, M = M)
```

```python


angles = np.arange(0, 180, 5)
Er = np.arange(0, maxEr(2, 40*931), maxEr(2, 40*931)/100)

fsize = 18

plt.subplots(figsize=(10, 6))
# plt.plot(angles, 1e9*thomsonE(thomsonCS(angles), 1, 40*931), label = "1 MeV")
plt.plot(thomsonE(angles, 1, 40*932) * 1e6, thomsonCS(angles), label = "1 MeV")
plt.plot(thomsonE(angles, 2, 40*932) * 1e6, thomsonCS(angles), label = "2 MeV")
plt.plot(thomsonE(angles, 3, 40*932) * 1e6, thomsonCS(angles), label = "3 MeV")

plt.legend(title = "Gamma energy")

plt.title("Thomson Spectrum", fontsize = fsize*1.2)
plt.ylabel(r"$ \frac{d\sigma}{dE_r} $ (AU)", fontsize = fsize)
plt.xlabel("Recoil energy (eV)", fontsize = fsize)
plt.show()
```
